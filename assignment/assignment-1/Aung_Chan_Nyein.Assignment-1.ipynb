## The myPOS tagset

myPOS v3.0 uses 15 POS tags.

| Tag | Meaning | Example | Gloss |
|---|---|---|---|
| `n` | Noun | ကျောင်း | school |
| `pron` | Pronoun | ကျွန်တော် | I (male speaker) |
| `v` | Verb | သွား | go |
| `adj` | Adjective | ကောင်း | good |
| `adv` | Adverb | မြန်မြန် | quickly |
| `part` | Particle | တယ် | statement particle |
| `ppm` | Post-positional marker | မှာ | at / in |
| `conj` | Conjunction | နှင့် | and |
| `int` | Interjection | အိုး | oh |
| `num` | Number (digit form) | ၂၀၂၆ | 2026 |
| `tn` | Number (spelled out) | တစ် | one |
| `abb` | Abbreviation | အ.စ.က | (org. abbreviation) |
| `fw` | Foreign word | အင်တာနက် | internet |
| `sb` | Symbol | % | percent sign |
| `punc` | Punctuation | ။ | full stop |


## Parsing word/tag sentences and making our own train/test split

````python
import math

def read_tagged_sentences(path):
    """Read 'word/tag word/tag ...' lines into a list of [(word, tag), ...] sentences."""
    sentences = []
    with open(path, encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            tokens = []
            for pair in line.split():
                if "/" not in pair:
                    continue
                word, tag = pair.rsplit("/", 1)
                if word and tag:
                    tokens.append((word, tag))
            if tokens:
                sentences.append(tokens)
    return sentences

def write_tagged_sentences(sentences, path):
    with open(path, "w", encoding="utf-8") as f:
        for sent in sentences:
            f.write(" ".join(f"{w}/{t}" for w, t in sent) + "\n")

sentences = read_tagged_sentences("data/mypos-ver.3.0.shuf.nopipe.txt")
print("Total tagged sentences:", len(sentences))

split_point = math.floor(len(sentences) * 0.8)
train_sentences = sentences[:split_point]
test_sentences = sentences[split_point:]
print("Train sentences:", len(train_sentences), " Test sentences:", len(test_sentences))

write_tagged_sentences(train_sentences, "data/train.postagged.txt")
write_tagged_sentences(test_sentences, "data/test.postagged.txt")
````

## Word-level feature extraction (`mypos_features.py`)

```
ကျွန်တော်/pron   ကျောင်း/n   သွား/v   တယ်/part   ။/punc
```

the third word (`သွား`, tag `v`, position i=2) produces:

```
v	w[-2]=ကျွန်တော်	w[-1]=ကျောင်း	w[0]=သွား	w[1]=တယ်	w[2]=။	w[-1]|w[0]=ကျောင်း|သွား	w[0]|w[1]=သွား|တယ်	pref1=သ	suff1=း	suff2=ား	is_digit=False	is_punc=False
```

and the first word (`ကျွန်တော်`, tag `pron`, i=0), where the window runs off the start of the
sentence:

```
pron	w[-2]=__BOS__	w[-1]=__BOS__	w[0]=ကျွန်တော်	w[1]=ကျောင်း	w[2]=သွား	w[-1]|w[0]=__BOS__|ကျွန်တော်	w[0]|w[1]=ကျွန်တော်|ကျောင်း	pref1=က	suff1=်	suff2=ာ်	is_digit=False	is_punc=False
```

````python
%%writefile mypos_features.py
#!/usr/bin/env python3
"""
mypos_features.py

Convert myPOS "word/tag" sentences into CRFsuite's native training format.

Input  (stdin): one sentence per line, tokens separated by spaces,
                each token written as word/tag, e.g.
                    ကျွန်တော်/pron ကျောင်း/n သွား/v တယ်/part ။/punc
Output (stdout): CRFsuite tab-separated format, one word per line,
                 blank line between sentences:
                    tag<TAB>feature<TAB>feature...

Usage:
    cat data/train.postagged.txt | python3 mypos_features.py > data/train.crfsuite.txt
"""

import sys

WINDOW = 2  # use w[-2] .. w[2]
MYANMAR_DIGITS = set("\u1040\u1041\u1042\u1043\u1044\u1045\u1046\u1047\u1048\u1049")
ARABIC_DIGITS = set("0123456789")
PUNC_CHARS = set("။၊()[]{}!?.,;:\"'")


def read_sentences(f):
    for line in f:
        line = line.strip()
        if not line:
            continue
        tokens = []
        for pair in line.split():
            if "/" not in pair:
                continue
            word, tag = pair.rsplit("/", 1)
            if word == "" or tag == "":
                continue
            tokens.append((word, tag))
        if tokens:
            yield tokens


def is_digit_word(w):
    return len(w) > 0 and all(ch in MYANMAR_DIGITS or ch in ARABIC_DIGITS for ch in w)


def is_punc_word(w):
    return len(w) > 0 and all(ch in PUNC_CHARS for ch in w)


def features_for_token(words, i):
    n = len(words)
    feats = []
    for offset in range(-WINDOW, WINDOW + 1):
        j = i + offset
        if j < 0:
            feats.append(f"w[{offset}]=__BOS__")
        elif j >= n:
            feats.append(f"w[{offset}]=__EOS__")
        else:
            feats.append(f"w[{offset}]={words[j]}")
    for offset in (-1, 0):
        j1, j2 = i + offset, i + offset + 1
        w1 = words[j1] if 0 <= j1 < n else ("__BOS__" if j1 < 0 else "__EOS__")
        w2 = words[j2] if 0 <= j2 < n else ("__BOS__" if j2 < 0 else "__EOS__")
        feats.append(f"w[{offset}]|w[{offset+1}]={w1}|{w2}")
    word = words[i]
    first_char = word[0] if word else ""
    last_char = word[-1] if word else ""
    last_two = word[-2:] if len(word) >= 2 else word
    feats.append(f"pref1={first_char}")
    feats.append(f"suff1={last_char}")
    feats.append(f"suff2={last_two}")
    feats.append(f"is_digit={is_digit_word(word)}")
    feats.append(f"is_punc={is_punc_word(word)}")
    return feats


def main():
    for tokens in read_sentences(sys.stdin):
        words = [w for w, t in tokens]
        tags = [t for w, t in tokens]
        for i in range(len(words)):
            row = [tags[i]] + features_for_token(words, i)
            print("\t".join(row))
        print()


if __name__ == "__main__":
    main()
````
