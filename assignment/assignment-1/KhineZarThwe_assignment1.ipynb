##Libraires Preparation

```python
# Create a smaller, memory-safe slice of your training file for Colab
input_file = '/home/khinezarthwe/word_segmentation/assignment/tag/train.crfsuite.txt'
small_file = '/home/khinezarthwe/word_segmentation/assignment/tag/train_small.crfsuite.txt'

max_lines = 15000  # Adjust this number if needed
line_count = 0

with open(input_file, 'r', encoding='utf-8') as fin, open(small_file, 'w', encoding='utf-8') as fout:
    for line in fin:
        fout.write(line)
        if line.strip() == '':
            line_count += 1
            if line_count >= max_lines:
                break

print(f"Created a smaller dataset slice with {line_count} sequences.")
```

```python
# Create a micro subset of 2,000 lines to prevent memory overflow crashes
input_file = '/home/khinezarthwe/word_segmentation/assignment/tag/train.crfsuite.txt'
micro_file = '/home/khinezarthwe/word_segmentation/assignment/tag/train_micro.crfsuite.txt'

max_lines = 2000
line_count = 0

with open(input_file, 'r', encoding='utf-8') as fin, open(micro_file, 'w', encoding='utf-8') as fout:
    for line in fin:
        fout.write(line)
        if line.strip() == '':
            line_count += 1
            if line_count >= max_lines:
                break

print(f"Micro dataset slice created with {line_count} sequences.")

# Train using the micro dataset
import pycrfsuite

trainer = pycrfsuite.Trainer(verbose=False)

def datagen(filepath):
    x_seq, y_seq = [], []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if not line:
                if x_seq:
                    yield x_seq, y_seq
                    x_seq, y_seq = [], []
                continue
            parts = line.split('\t')
            y_seq.append(parts[-1])
            x_seq.append(parts[:-1])
    if x_seq:
        yield x_seq, y_seq

for xseq, yseq in datagen(micro_file):
    trainer.append(xseq, yseq)

trainer.set_params({
    'c1': 1.0,
    'c2': 1e-3,
    'max_iterations': 10,
    'feature.minfreq': 3
})

print("Training micro model...")
trainer.train('ws_tag3.model')
print("Model training successfully completed and saved as ws_tag3.model!")
```
