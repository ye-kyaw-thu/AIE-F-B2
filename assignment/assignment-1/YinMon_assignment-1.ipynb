**Convert data into 'word/tag'**

```python
def parse_pos_line(line):
    sentence = []
    for chunk in line.strip().split():
        for token in chunk.split('|'):
            if not token or '/' not in token:
                continue
            word, tag = token.rsplit('/', 1)
            sentence.append((word, tag))
    return sentence

def parse_pos_file(path):
    sentences = []
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                sentences.append(parse_pos_line(line))
    return sentences

train_sentences = parse_pos_file('/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/train.mypos.txt')
test_sentences = parse_pos_file('/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/test.mypos.txt')
train_sentences[0]
```

**Convert tagged sentences into CRFsuite format**

```python
def escape(feature):
    return feature.replace(':', '__COLON__')

def to_crfsuite_format(sentences, window=2):
    """Convert (word, tag) sentences into CRFsuite training-format lines:
    label<TAB>w[-n..n]=...<TAB>w[i]|w[i+1]=...<TAB>w[i]|w[i+1]|w[i+2]=...<TAB>__BOS__/__EOS__"""
    lines = []
    for sentence in sentences:
        words = [w for w, _ in sentence]
        tags = [t for _, t in sentence]
        n = len(words)
        for i in range(n):
            feats = []
            for offset in range(-window, window + 1):
                j = i + offset
                if 0 <= j < n:
                    feats.append(f"w[{offset}]={words[j]}")
            for offset in range(-window, window):
                j1, j2 = i + offset, i + offset + 1
                if 0 <= j1 < n and 0 <= j2 < n:
                    feats.append(f"w[{offset}]|w[{offset + 1}]={words[j1]}|{words[j2]}")
            for offset in range(-window, window - 1):
                j1, j2, j3 = i + offset, i + offset + 1, i + offset + 2
                if 0 <= j1 < n and 0 <= j3 < n:
                    feats.append(f"w[{offset}]|w[{offset + 1}]|w[{offset + 2}]={words[j1]}|{words[j2]}|{words[j3]}")
            if i == 0:
                feats.append('__BOS__')
            if i == n - 1:
                feats.append('__EOS__')
            lines.append(tags[i] + '\t' + '\t'.join(escape(feat) for feat in feats))
        lines.append('')
    return lines

def write_crfsuite_file(sentences, output_path, window=2):
    with open(output_path, 'w', encoding='utf-8') as f:
        for line in to_crfsuite_format(sentences, window=window):
            f.write(line + '\n')

write_crfsuite_file(train_sentences, '/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/train.crfsuite.txt')
write_crfsuite_file(test_sentences, '/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/test.crfsuite.txt')
```

```python
def write_tagged_file(sentences, output_path):
    with open(output_path, 'w', encoding='utf-8') as f:
        for sentence in sentences:
            for word, tag in sentence:
                f.write(f"{word}\t{tag}\n")
            f.write('\n')

write_tagged_file(train_sentences, '/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/train.wtag.txt')
write_tagged_file(test_sentences, '/Users/mon/Desktop/Project/Python/AI_Proj/myPOS/pos_tag/test.wtag.txt')
```
