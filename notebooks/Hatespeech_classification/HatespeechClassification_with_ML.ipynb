## Summary

Across all five classical machine-learning baselines evaluated on the **myHateSpeech** dataset (9,137 training rows / 1,000 test rows, 9 imbalanced hate-speech classes), the overall test accuracy clustered in a narrow band between **0.51 and 0.60**, with **SGD using unigram TF-IDF features achieving the best raw accuracy (0.600)**, followed closely by **SVM (0.597)** and **Random Forest (0.592)** on the same feature representation. Decision Tree was clearly the weakest model family — its ~0.92 training score against ~0.51–0.52 test score is textbook overfitting, and even its best test configuration (0.522) trails the other learners by 6–8 accuracy points. Random Forest, SGD, and SVM all produced similar peak results within ~1 point of each other, which suggests the ceiling here is set by the data rather than the classifier: the rare hate-speech categories (`ra` = 105, `re` = 90, `ed` = 206 in training) are almost never predicted, driving macro-F1 down to ~0.25 everywhere despite weighted-F1 around 0.51–0.53. TF-IDF representations consistently edged out raw counts for the linear models (SGD, SVM), while Random Forest was more indifferent between them. The two-phase hyperparameter tuning of SGD (best: `squared_hinge` loss, `adaptive` learning rate, `eta0≈0.0032`, `elasticnet` penalty, `alpha≈7.4e-5`) **did not improve** on the untuned SGD — test accuracy actually dropped from 0.600 to 0.583 — indicating that the default `SGDClassifier` parameters were already near-optimal for this dataset and that the bottleneck is the class imbalance across the 9 hate-speech categories, not the hyperparameters. If higher accuracy is the goal, the next steps should focus on data-level techniques (class weighting, resampling, or merging rare hate-speech classes) rather than further model tuning.
