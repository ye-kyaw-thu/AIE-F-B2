## Training/Testing inside Python Code

````python
import fasttext
model = fasttext.train_supervised(input="./data/ltrain.txt", lr=1.0, epoch=25)
````

````python
model.test("./data/ltest.txt")
````

````python
model = fasttext.train_supervised(input="./data/ltrain.txt", lr=1.0, epoch=25, wordNgrams=3, bucket=200000, dim=150, loss='hs')
````

````python
model.test("./data/ltest.txt")
````
