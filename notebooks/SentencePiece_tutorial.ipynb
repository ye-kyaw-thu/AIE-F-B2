## Sub-word Segmentation with Small Model

````python
%pwd
````

````python
# 2. Load the trained model.
small_model = spm.SentencePieceProcessor(model_file='small.model')
````

````python


# 3. Encode raw text into subword pieces (strings) or vocabulary IDs (integers).
text = "စာမေးပွဲမေးခွန်း ပေါက်ကြားမှုတွေ အကြိမ်ကြိမ် ဖြစ်ပွားခဲ့တာအပေါ် ဒေါသထွက်မှုကနေ ပေါ်ပေါက်လာတဲ့ လူငယ်တွေဦးဆောင်တဲ့ လှုပ်ရှားမှုဖြစ်တဲ့ ပိုးဟပ်ပါတီ- Cockroach Janta Party (CJP) ဟာ နှစ်ပေါင်းများစွာအတွင်း အိန္ဒိယနိုင်ငံမှာ အကြီးမားဆုံး ကျောင်းသားလှုပ်ရှားမှုတစ်ခုအဖြစ် လျင်မြန်စွာ ကြီးထွားလာခဲ့ပါတယ်။"
pieces = small_model.encode(text, out_type=str)
ids = small_model.encode(text, out_type=int)

print(f"Pieces: {pieces}")
````

````python
print(f"IDs:    {ids}")
````

````python
# 4. Decode IDs or pieces back into the original text.
# The reconstruction is completely lossless and reversible!
print(small_model.decode(ids))
````

````python
print(small_model.decode(pieces))
````
