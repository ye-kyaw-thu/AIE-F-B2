## Train minbpe Tokenizer

```python
import os
import time


# Tell Python where to find the cloned minbpe folder
sys.path.append(os.path.abspath("./minbpe"))

from minbpe import BasicTokenizer, RegexTokenizer

# open some text and train a vocab of 512 tokens
text = open("/home/ye/aif2/word-seg/data/myPOS/otest.txt", "r", encoding="utf-8").read()

# create a directory for models, so we don't pollute the current directory
os.makedirs("models", exist_ok=True)

t0 = time.time()
for TokenizerClass, name in zip([BasicTokenizer, RegexTokenizer], ["basic", "regex"]):

    # construct the Tokenizer object and kick off verbose training
    tokenizer = TokenizerClass()
    tokenizer.train(text, 512, verbose=True)
    # writes two files in the models directory: name.model, and name.vocab
    prefix = os.path.join("models", name)
    tokenizer.save(prefix)
t1 = time.time()

print(f"Training took {t1 - t0:.2f} seconds")
```

## UTF-8 and Multi-Byte Characters

အင်္ဂလိပ်စာမှာက letter တိုင်းက single byte (i.e. ASCII 0-127) နဲ့ အဆင်ပြေတယ်။ ဥပမာ 'e' က byte '101'။  
ဒါပေမဲ့ UTF-8 မြန်မာစာမှာက character တစ်လုံးကို 3 bytes နဲ့ ကိုယ်စားပြုတယ်။ မြန်မာစာလုံးတိုင်းက prefix bytes '\xe1\x80' နဲ့စတယ်။  

## Step-by-Step Line Explanation

**Merge 1:** tokens (225, 128) -> ID 256 ([Partial bytes: b'\xe1\x80'])
Algorithm က byte 225 (\xe1) နဲ့ byte 128 (\x80) ကို တွဲလျက် ခဏခဏတွေ့ရတယ်။ တခြား အတွဲတွေထက်ကို ဒီအတွဲကို ပိုတွေ့ရတာမို့ အရင်ဆုံး စပြီး merge လုပ်တယ်။ တနည်းအားဖြင့် BPE စလုပ်ပါတယ်။  

**This is the foundational building block for every Myanmar letter.**  
ဒီ ID 256 က မြန်မာစာရဲ့ ဗျည်းတို့ သရတို့ စတဲ့ စာလုံးတွေကို ကိုယ်စားပြုတာ မဟုတ်ပေမဲ့၊ မြန်မာစာစာလုံးတိုင်းမှာ ရှေ့ကနေ ပါနေတဲ့ 2-byte prefix (\xe1\x80) ကို ကိုယ်စားပြုပါတယ်။ 

**Merge 2:** tokens (32, 256) -> ID 257 ([Partial bytes: b' \xe1\x80'])
ဒီmerge ကကျတော့ ID 256 (the prefix \xe1\x80) နဲ့Byte 32 (a space ' ') အတွဲအတွက် BPE လုပ်သွားတာပါ။  

**This shows how BPE handles word boundaries.**   
ဒါကဘာကိုဆိုလိုတာလဲ ဆိုတော့ algorithm က မြန်မာစာစာလုံး သို့မဟုတ် ဝဏ္ဏတွေရဲ့ ရှေ့မှာ မကြာခဏ space ရှိတတ်တယ်ဆိုတာကို ဂရုပြုမိတယ်လို့ နားလည်လို့ ရပါတယ်။  

*Merges 3 to 8: ဒီနေရာကနေ စပြီး ပုံမှန်မြန်မာစာ စာလုံးတွေဖြစ်တဲ့ သရတို့ အသတ်တို့ကို စတင်မြင်ရပါပြီ။  

**Merge 3:** (256, 186) -> ID 258 ('်')

**Merge 4:** (256, 172) -> ID 259 ('ာ')

**Merge 5:** (256, 184) -> ID 260 ('း')

**Merge 6:** (256, 175) -> ID 261 ('ု')

**Merge 7:** (256, 177) -> ID 262 ('ေ')

**Merge 8:** (256, 173) -> ID 263 ('ိ')

## Testing minbpe Model

```python
import os
import sys

# Ensure Python knows where to find the minbpe library
sys.path.append(os.path.abspath("./minbpe"))

from minbpe import BasicTokenizer, RegexTokenizer

# --- CHOOSE WHICH MODEL TO LOAD ---
# You can change this to "regex" if you want to test the RegexTokenizer model
model_type = "basic" 
prefix = os.path.join("models", model_type)

# 1. Instantiate the appropriate Tokenizer class
if model_type == "regex":
    tokenizer = RegexTokenizer()
else:
    tokenizer = BasicTokenizer()

# 2. Load the trained model files (this looks for models/basic.model and models/basic.vocab)
tokenizer.load(prefix + ".model")
print(f"Successfully loaded '{model_type}' tokenizer! Vocabulary size: {len(tokenizer.vocab)}")

# 3. Test encoding and decoding
test_text = "အိန္ဒိယကျောင်းသားလှုပ်ရှားမှု"
encoded_tokens = tokenizer.encode(test_text)
print("\nEncoded IDs:", encoded_tokens)

decoded_text = tokenizer.decode(encoded_tokens)
print("Decoded Text:", decoded_text)

# Optional: Verify it's a lossless round-trip
assert test_text == decoded_text, "Decoded text does not match original!"
print("Round-trip test passed successfully!")
```

## Production Byte-Level BPE (BBPE) with Hugging Face 

Link: https://github.com/huggingface/tokenizers

```python
from tokenizers import Tokenizer, decoders
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import ByteLevel

# 1. Initialize a Byte-Level BPE model
tokenizer = Tokenizer(BPE(unk_token="<unk>"))

# 2. IMPORTANT: Set BOTH the Pre-Tokenizer AND the Decoder for Byte-Level mapping
tokenizer.pre_tokenizer = ByteLevel(add_prefix_space=False)
tokenizer.decoder = decoders.ByteLevel()

# 3. Setup the trainer with a target vocabulary size
trainer = BpeTrainer(
    vocab_size=1000, 
    special_tokens=["<unk>"]
)

# 4. Train on your Myanmar text file
files = ["/home/ye/aif2/word-seg/sub-word/big_data/bbc-cockroach-news.txt"]
tokenizer.train(files, trainer)

# 5. Encode text and look at the behavior
output = tokenizer.encode("အိန္ဒိယကျောင်းသားလှုပ်ရှားမှု")
print("Tokens:", output.tokens)
print("IDs:", output.ids)

# 6. Decode back using the proper decoder
decoded_text = tokenizer.decode(output.ids)
print("Decoded Text:", decoded_text)
```

## For Your Information

Hugging Face's ByteLevel pre-tokenizer uses a clever trick invented for GPT-2: it maps raw bytes (0–255) to a specific set of visible, printable Unicode characters so that control characters, spaces, and binary data don't break string formatting or tokenization files.  

A space ( ) is mapped to the character Ġ (U+0120).  

Raw bytes of Myanmar letters (like \xe1\x80) are mapped to printable ASCII/Latin equivalents like áĢ¡ during the intermediate tokenization steps.  

## High-Speed Production Tokenization with tiktoken

Demonstrate how modern LLMs (like GPT-4) use fixed pre-trained.  

```python
!pip install tiktoken --break-system-packages
```

```python
import tiktoken

# OpenAI doesn't natively have a pre-trained Myanmar-only tiktoken encoder out-of-the-box, 
# but we can inspect how cl100k_base (GPT-4 encoding) handles Myanmar text via bytes.
enc = tiktoken.get_encoding("cl100k_base")

text_sample = "အိန္ဒိယနိုင်ငံမှာ အကြီးမားဆုံး ကျောင်းသားလှုပ်ရှားမှု"
encoded = enc.encode(text_sample)

print("GPT-4 Token IDs for Myanmar text:", encoded)
print("Number of tokens:", len(encoded))

# Decode back to verify lossless round-trip
print("Decoded back:", enc.decode(encoded))
```

## What is cl100k_base?  

- cl100k_base မော်ဒယ်က OpenAI က သူ့ရဲ့ မော်ဒယ်အများစုအတွက် သုံးနေတဲ့ pre-trained encoding မော်ဒယ်ပါပဲ။  
- cl (ChatGPT/Clrir) က GPT-4 မော်ဒယ်လောက်ကနေ စထွက်လာတဲ့ မော်ဒယ်တွေကို ညွှန်းပါတယ်။  
- base ဆိုတာက task-specific fine-tuning မတိုင်ခင် အခြေခံ layer ကို ဆိုလိုတာပါ။  
- GPT-4, GPT-4o, and GPT-3.5-Turbo တွေမှာ default tokenizer အဖြစ်သုံးတဲ့ မော်ဒယ်ပါ။
