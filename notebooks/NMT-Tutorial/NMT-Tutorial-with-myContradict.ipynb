```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
clean-punctuation.py
Cleaning punctuation characters for Burmese or Myanmar language.
Python port of clean-punctuation.pl (Ye Kyaw Thu, 22 May 2022)

Usage:
    python3 clean-punctuation.py <input-file> > <output-file>
    cat input.txt | python3 clean-punctuation.py > output.txt
"""

import sys
import re

# Punctuation characters to remove (same set as the Perl script)
PUNCT_RE = re.compile(r'[၊။\-()\[\]/\\!?,"<>]')
MULTI_SPACE_RE = re.compile(r' +')


def clean_line(line: str) -> str:
    """Remove target punctuation, trim, and collapse multiple spaces."""
    line = PUNCT_RE.sub('', line)
    line = line.strip()
    line = MULTI_SPACE_RE.sub(' ', line)
    return line


def main() -> None:
    # Read from file argument, or fall back to stdin
    if len(sys.argv) >= 2:
        fin = open(sys.argv[1], 'r', encoding='utf-8')
        close_fin = True
    else:
        fin = sys.stdin
        close_fin = False

    try:
        for raw in fin:
            line = raw.rstrip('\n').rstrip('\r')
            if not line.strip():          # skip empty / whitespace-only lines
                continue
            cleaned = clean_line(line)
            if cleaned:                   # skip lines that became empty
                print(cleaned)
    finally:
        if close_fin:
            fin.close()


if __name__ == '__main__':
    main()
```

## Prepare Bash Shell Script for Transformer Model Building

````python
%cd /mnt/disk1/ye/exp/contra/
````

```bash
#!/bin/bash

## Written by Ye Kyaw Thu, Language Understanding Lab, AI Mario, Myanmar
## Experiments for AI Engineering Fundamental Class (Batch-2)
## Date: 26 Sept 2026

## Old Notes
#     --mini-batch-fit -w 10000 --maxi-batch 1000 \
#    --mini-batch-fit -w 1000 --maxi-batch 100 \
#     --tied-embeddings-all \
#     --tied-embeddings \
#     --valid-metrics cross-entropy perplexity translation bleu \
#     --transformer-dropout 0.1 --label-smoothing 0.1 \
#     --learn-rate 0.0003 --lr-warmup 16000 --lr-decay-inv-sqrt 16000 --lr-report \
#     --optimizer-params 0.9 0.98 1e-09 --clip-norm 5 \

model_folder="model.tf.syl";
mkdir -p ${model_folder};
data_path="/mnt/disk1/ye/exp/contra/data/syl/";
vocab_path="/mnt/disk1/ye/exp/contra/data/vocab/"
src="src"; tgt="tgt";

marian \
    --model ${model_folder}/model.npz --type transformer \
    --train-sets ${data_path}/train.${src} ${data_path}/train.${tgt} \
    --max-length 200 \
    --vocabs  ${vocab_path}/word_vocab.yml  ${vocab_path}/word_vocab.yml \
    --mini-batch-fit -w 1000 --maxi-batch 100 \
    --early-stopping 10 \
    --valid-freq 5000 --save-freq 5000 --disp-freq 500 \
    --valid-metrics cross-entropy perplexity bleu \
    --valid-sets ${data_path}/dev.${src} ${data_path}/dev.${tgt} \
    --valid-translation-output ${model_folder}/valid.${src}-${tgt}.output \
    --quiet-translation \
    --valid-mini-batch 128 \
    --beam-size 6 --normalize 0.6 \
    --log ${model_folder}/train.log \
    --valid-log ${model_folder}/valid.log \
    --enc-depth 2 --dec-depth 2 \
    --transformer-heads 8 \
    --transformer-postprocess-emb d \
    --transformer-postprocess dan \
    --transformer-dropout 0.3 --label-smoothing 0.1 \
    --learn-rate 0.0003 --lr-warmup 0 --lr-decay-inv-sqrt 16000 --lr-report \
    --clip-norm 5 \
    --tied-embeddings \
    --devices 0 --sync-sgd --seed 1111 \
    --exponential-smoothing \
    --dump-config > ${model_folder}/${src}-${tgt}.config.yml

time marian -c ${model_folder}/${src}-${tgt}.config.yml 2>&1 | tee ${model_folder}/tf.${src}-${tgt}.log
```
