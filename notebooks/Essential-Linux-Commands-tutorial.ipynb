ဖိုင်တစ်ဖိုင်ရဲ့ date/time information ကို ပြင်ချင်ရင်လည်း touch command ကို သုံးကြတယ်။  

````python
!touch -m newfile.txt
````

The `-t` option takes a timestamp in the format `[[CC]YY]MMDDhhmm[.ss]`, where `CC` is the century, `YY` is the year, `MM` is the month, `DD` is the day, `hh` is the hour, `mm` is the minute, and `ss` is the second.

````python
!touch -t 201011131030.00 newfile.txt
````

````python
!ls -lh newfile.txt
````

အထက်မှာ မြင်ရတဲ့အတိုင်း newfile.txt ဖိုင်ရဲ့ creation data/time က 2010 Nov 13 ရက်နေ့အဖြစ် ပြောင်းလဲသွားတာကို တွေ့ရပါလိမ့်မယ်။

**တစ်ခု သတိထားရမှာက locate command ကို terminal မှာ run တာက မြန်ပေမဲ့ Jupyter Notebook (သို့) Jupyter Lab ရဲ့ cell အတွင်းက run တဲ့အခါမှာ တအားကြာနေတာမျိုး ဖြစ်တတ်ပါတယ်။**

အခြေခံအားဖြင့်က `locate` command က သတ်မှတ်ထားတဲ့ database ထဲကနေ ရှာဖွေပေးတာပါ။ အဲဒါကြောင့် မြန်ပါတယ်။ သို့သော် အဲဒီ database ကိုတော့ update လုပ်ပေးဖို့ လိုအပ်ပါတယ်။

ဖိုင်နှစ်ဖိုင်ကို တူသလား၊ မတူဘူးလား ဆိုတာကို စာကြောင်း တစ်ကြောင်းချင်းစီ တိုက်စစ်ဖို့အတွက် အသုံးပြုတဲ့ command ပါ။  

````python
!diff ./researcher.txt ./professor.txt
````

**`1c1,3` ဆိုတာဘာလဲ။**
    
`diff` ရဲ့ output မှာ `1c1,3` ဆိုတာက ပထမဖိုင်ကို ဒုတိယဖိုင်နှင့် တူညီအောင် ဘယ်လိုပြင်ရမလဲ ဆိုတာကို ပြောပြတာပါ။
*   **`1`**: ပထမဖိုင် (`researcher.txt`) ရဲ့ စာကြောင်း ၁ ကို ရည်ညွှန်းတယ်။
*   **`c`**: **"change"** (အစားထိုး/ပြောင်းလဲရန်) ဆိုတဲ့ အဓိပ္ပာယ်ရပါတယ်။ (အခြားအက္ခရာများဖြစ်တဲ့ `a` က ထည့်ရန်၊ `d` ဆိုတာက ဖျက်ရန်)။
*   **`1,3`**: ဒုတိယဖိုင် (`professor.txt`) ရဲ့ စာကြောင်း ၁ မှ စာကြောင်း ၃ အထိကို ရည်ညွှန်းပါတယ်။  
*အဓိပ္ပာယ်က -* "ပထမဖိုင်ရဲ့ စာကြောင်း ၁ ကို ဒုတိယဖိုင်ရဲ့ စာကြောင်း ၁ မှ ၃ အထိ အစားထိုးပါ" လို့ ဆိုလိုပါတယ်။


````python
!diff -u ./researcher.txt ./professor.txt
````

**`-`, `+` နဲ့ `@@ -1 +1,3 @@` တို့ရဲ့ အဓိပ္ပာယ်များ**
Unified (`-u`)  - ရဲ့ output မှာ ပြနေတာတွေက  
*   **`-` (အနုတ်)။** ပထမဖိုင် (မူလ) တွင်ရှိပြီး ဖျက်ရမယ့် (သို့) အစားထိုးခံရမယ့် စာကြောင်းကို ပြတာ။
*   **`+` (အပေါင်း)။** ဒုတိယဖိုင် (အသစ်) တွင် ထည့်သွင်း (သို့) ပေါင်းထည့် ထားတဲ့ စာကြောင်းကို ပြတာ။
*   **`@@ -1 +1,3 @@`:** ဆိုတာက ပြောင်းလဲမှုအပိုင်းအခြေအနေ (hunk header) ကို ပြောပြနေတာ ဖြစ်တယ်။
    *   **`-1`**: မူလဖိုင်မှာ စာကြောင်း ၁ မှစတင်ပြီး စာကြောင်း ၁ ကြောင်း ပါဝင်သော ပြောင်းလဲမှုဖြစ်တယ်ဆိုတာကို ပြတာ။
    *   **`+1,3`**: ဖိုင်အသစ်မှာ စာကြောင်း ၁ မှစတင်ပြီး စာကြောင်း ၃ ကြောင်း ပါဝင်ကြောင်း ပြသတာ။

### 65. sha256sum

`md5sum` လိုပါပဲ။ ဖိုင်တစ်ဖိုင်အတွက် checksum value တွက်ထုတ်ပေးတဲ့ command ပါ။ ဒီ `sha256sum` ကတော့ ပိုရှည်လျားတဲ့ checksum တန်ဖိုးကို တွက်ထုတ်ပေးနိုင်ပါတယ်။ Security အရ ပိုပြီးတော့ကောင်းတဲ့ file fingerprint ကို ထုတ်ပေးနိုင်ပါတယ်။  

အနည်းငယ် ဖြည့်စွက်ရှင်းပြရရင် `md5sum` ရော `sha256sum` ရော နှစ်မျိုးစလုံးက cryptographic hash function တွေပါပဲ။ ဖိုင်တစ်ဖိုင်ရဲ့ အရွယ်အစားက ဘယ်လောက်ပဲ ကြီးကြီး fixed-length string (i.e. fingerprint) ထုတ်ပေးနိုင်ပါတယ်။ `md5sum` command ကတော့ 128 bits (32 character string) အနေနဲ့ထုတ်ပေးမှာဖြစ်ပြီး၊ `sha256sum` command ကတော့ 256 bits (64 characters string) အနေနဲ့ ထုတ်ပေးမှာ ဖြစ်ပါတယ်။ တွက်တာက အခြေခံအားဖြင့် bitwise rotation နဲ့ bitwise addition တွေကို အခြေခံထားပါတယ်။ အဲဒါကြောင့် ဖိုင်ထဲမှာ အပြောင်းအလဲ တစ်ခုခု လုပ်ထားရင် (သို့) အပြောင်းအလဲ အနည်းငယ် ဖြစ်သွားရင်တောင် သိနိုင်ပါတယ်။ Avalanche effect ကိုသိအောင် engineering လုပ်ထားတာမို့ပါ။  


````python
!sha256sum ./family.txt 
````

````python
!sha256sum ./family.txt > ./family.txt.sha
````

````python
!cat ./family.txt.sha
````

````python
!sha256sum --help
````

`-c` option ကိုသုံးပြီး check လုပ်ကြည့်ရအောင်။

#### Extract All Unique Skills into a Single Flat List (flatten and unique)

Json ဖိုင်ထဲမှာ ရှိတဲ့ user skills တွေအားလုံးကို စုလိုက်ပြီး flatten unique list တစ်ခုအဖြစ် လုပ်ချင်တယ်ဆိုပါစို့...

#### Transform the Array into a Lookup Dictionary/Object (INDEX)

တကယ်တမ်းက array index ထောက်တာမျိုးကို မသုံးပဲ INDEX ဆိုတဲ့ function ကိုခေါ်သုံးပြီးလည်း key ပေးပြီး ဆွဲထုတ်လို့ ရပါသေးတယ်။  

````python
!jq 'INDEX(.id)' users.json
````

````python
!jq -r '.[] | [.id, .name, .active] | @tsv' users.json
````

    - PID: 2

    - PID: 3

    - Command: \_ [pool_workqueue_release]
    - Role: This is a kernel worker thread spawned by kthreadd (shown visually with the \_ branch right underneath it).

#### Understanding the Layout  

`top` ရဲ့ interface က section နှစ်ခုအနေနဲ့ ရှိပါတယ်။ အပေါ်ဆုံး အပိုင်းက summary လုပ်ပြတာဖြစ်ပြီး အောက်ဘက်က ဇယားကတော့ process table ပါ။

1. **The Header (Top Summary)**:
* **Uptime & Load Average**: Shows how long the system has been running and the CPU load over 1, 5, and 15-minute intervals.
* **Tasks**: Total running, sleeping, or zombie processes.
* **CPU (`%Cpu(s)`)**: Real-time breakdown of user usage (`us`), system/kernel usage (`sy`), and idle time (`id`).
* **Memory (`MiB Mem / Swap`)**: Total, free, used, and buffered/cached RAM.

2. **The Process Table (Bottom Grid)**:
* Columns like **`PID`** (Process ID), **`USER`** (Owner), **`%CPU`**, **`%MEM`**, and **`COMMAND`**.


#### Essential Interactive Keys (While `top` is running)

1. Sort Processes by CPU Usage (`Shift + P`)
2. Toggle Individual CPU Cores (`1`)
3. Filter Processes by a Specific User (`u`)
4. Kill a Misbehaving Process (`k`)
5. Toggle Full Command Paths (`c`)
6. Help (`h`)

#### `top` vs. `htop` Comparison

| Feature | `top` | `htop` |
| --- | --- | --- |
| **Interface** | Plain text, basic layout, monochrome by default. | Colorful, visual meters for CPU/Memory, sleek ASCII layout. |
| **Mouse Support** | No. Must use keyboard shortcuts only. | **Yes!** You can click rows, scroll, and click buttons. |
| **Killing Processes** | Requires typing the `PID` manually (`k` -> enter PID). | Select the process with arrow keys/mouse and press `F9`. |
| **Searching/Filtering** | Requires external commands or complex hotkeys. | Press `F4` to type and filter processes instantly. |
| **Default Availability** | Pre-installed on virtually every Linux distribution. | Often requires a quick manual install (`sudo apt install htop`). |

### 70. kill

Running process တွေကို ရပ်ပေးတဲ့ command ပါ။ `kill` ဆိုတဲ့ ခပ်ကြမ်းကြမ်းနာမည်ကို သုံးထားပေမဲ့ တကယ်က running process ကို အော်တို တန်းရပ်ချပစ်လိုက်တာမျိုးတော့ မဟုတ်ပါဘူး။ သူက run လုပ်နေတဲ့ process ဆီကို signal ပို့ပြီး ရပ်ခိုင်းတဲ့ ပုံစံပါ။  ပြီးတော့ `kill` command သုံးပြီး အဲဒီ ရပ်သွားတဲ့ process ကိုလည်း `pause`, `resume`, (သို့) `reload` လုပ်တာမျိုး လုပ်လို့ရပါတယ်။   

````python
!whatis kill
````

````python
!kill --help
````

#### Gracefully Terminate a Process (Default Signal: SIGTERM / 15)

ဥပမာ process ID နံပါတ် 12345 ကို ရပ်ချင်ရင် အောက်ပါအတိုင်း command ပေးပါ။  

kill 12345

#### Forcefully Kill a Stubborn Process (Signal: SIGKILL / 9)

ရပ်ခိုင်းလို့ process က မရပ်ပေးနိုင်တာမျိုးဖြစ်နေရင် (ဥပမာ frozen state ဖြစ်နေတာမျိုး)၊ Linux kernel ကို by force နဲ့ ရပ်ခိုင်းလို့ ရပါတယ်။ အဲဒီလိုမျိုး လုပ်မယ်ဆိုရင်တော့ အောက်ပါ command တွေကို အသုံးပြုပါ။  

`kill -9 12345`
# Alternatively, using the signal name:
`kill -s SIGKILL 12345`

#### List All Available Signals (-l)

တကယ်ကတော့ Linux process တွေက လက်ခံနိုင်တဲ့ signal တွေ အမျိုးမျိုးရှိပါတယ်။ အဲဒီ signal list အားလုံးကို လေ့လာချင်ရင်တော့ အောက်ပါအတိုင်း `-l` option သုံးပြီး command ပေးကြည့်ပါ။  


````python
!kill -l
````

#### Terminate a Process by Name instead of PID (pkill)

PID မသုံးပဲ Process Name ကိုပြောပြီးလည်း kill လုပ်လို့ ရပါတယ်။  
ဥပမာ အနေနဲ့ python3 process ကို ရပ်ခိုင်းချင်ရင်...  

`pkill python3`

````python
!pkill --help
````

#### Gracefully Stop All Processes Matching a Name (killall)  

Running process တွေကို kill လုပ်တာနဲ့ ဆက်စပ်ပြီး နောက်ထပ် အသုံးဝင်တဲ့ command တစ်ခုကို မိတ်ဆက်ပေးရရင်တော့ `killall` ပါ။  
တချို့ process တွေက run လိုက်ရင် process တစ်ခုတည်းအနေနဲ့ running လုပ်တာမဟုတ်ပဲ multi-process အနေနဲ့ ခွဲပြီး run တာမျိုးတွေလည်း လုပ်တတ်ပါတယ်။ ဥပမာ `nginx` လို process မျိုးပါ။ သူက web traffic (သို့) client request တွေနဲ့ ဆိုင်တဲ့ process ပါ။ လက်ရှိ စက်မှာ ရှာကြည့်ချင်ရင် အောက်ပါလိုမျိုး command ပေးပြီး ရှာဖွေကြည့်လို့ ရနိုင်ပါတယ်။  


````python
!ps -ef | grep nginx
````

အထက်မှာ မြင်ရတဲ့အတိုင်းပါ။ `master process` နဲ့ `worker process` ဆိုပြီး အများကြီး ခွဲပြီး running လုပ်နေတာကို တွေ့ကြရပါလိမ့်မယ်။  
အဲဒီလိုမျိုး run နေတဲ့ `nginx` ရဲ့ multiple instances တွေအားလုံးကို ရပ်ခိုင်းချင်ရင်တော့ အောက်ပါလိုမျိုး command ပေးပါ။  

`killall nginx`

တကယ် တတ်ချင်ရင် command တွေကိုလည်း သေသေချာချာ သုံးတတ်အောင် လေ့လာပြီး၊ bash, zsh, csh တို့ထဲက shell တစ်မျိုးမျိုးကို ကျွမ်းကျင်ပိုင်နိုင်အောင် ပြင်ဆင်ထားစေချင်ပါတယ်။  

```
Download → unzip → extract JSON text → clean HTML → lowercase → shuffle → shard → sanity-check the line count– eight commands from today’s talk, chained into one real workflow.
```

```bash
wget-c https://example.com/reviews.zip
unzip reviews.zip && rm reviews.zip
jq-r '.[] | .text' reviews.json > reviews.txt
sed-i 's/<br \/>/ /g' reviews.txt
tr '[:upper:]' '[:lower:]' < reviews.txt > reviews_clean.txt
shuf reviews_clean.txt-o reviews_shuffled.txt
split-n 3-d reviews_shuffled.txt part_
wc-l reviews_shuffled.txt
```
